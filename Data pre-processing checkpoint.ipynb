```python
# Missing values in counter_number by counter_type
missing_counter_by_type = df[df['counter_number'].isnull()]['counter_type'].value_counts()
print(missing_counter_by_type)

# Missing values in reading_remarque by counter_type
missing_reading_by_type = df[df['reading_remarque'].isnull()]['counter_type'].value_counts()
print(missing_reading_by_type)
```

```python
# Check the relationship between missing values in 'counter_number' and 'client_id'
missing_counter_number = df[df['counter_number'].isnull()]

# Group by 'client_id' and count missing values
client_missing_counter = missing_counter_number.groupby('client_id').size()

client_missing_counter
```

```python
# Missing values in reading_remarque
missing_reading_remarque = df[df['reading_remarque'].isnull()]
missing_by_client_reading = missing_reading_remarque['client_id'].value_counts()
missing_by_client_reading
```

```python
# Let's group by 'client_id' and calculate aggregates for 'counter_number' to see the distrubution
client_counter_number_ranges = df.groupby('client_id')['counter_number'].agg(['min', 'max', 'mean', 'median']).reset_index()

client_counter_number_ranges

# Run a descriptive statistics on it
client_counter_number_ranges.describe()
```

```python
# Calculate mean for each client 
client_counter_number_ranges = df.groupby('client_id')['counter_number'].agg(['mean']).reset_index()

# Create a dictionary mapping client_id to their mean counter_number
client_counter_number_dict = client_counter_number_ranges.set_index('client_id')['mean'].to_dict()

# Replace missing values in 'counter_number' with the corresponding client's mean
df['counter_number'] = df['counter_number'].fillna(df['client_id'].map(client_counter_number_dict))

# Verify that there are no more missing values in 'counter_number'
df['counter_number'].isnull().sum()
```

```python
# Group by 'client_id' and calculate aggregates for 'reading_remarque' to see the distrubution
client_reading_remarque_ranges = df.groupby('client_id')['reading_remarque'].agg(['min', 'max', 'mean', 'median']).reset_index()
client_reading_remarque_ranges

# Run descriptive statistics
client_reading_remarque_ranges.describe()
```

```python
# Replace missing values in 'reading_remarque' with the corresponding client's mean
df['reading_remarque'] = df.groupby('client_id')['reading_remarque'].transform(
    lambda x: x.fillna(x.mean())
)

# Verify that there are no more missing values in 'reading_remarque'
df['reading_remarque'].isnull().sum()
```
