```python
import pandas as pd

data = {'Name': ['John', 'Mary', 'Bob', 'Sarah', 'Tom', 'Lisa'], 
        'Department': ['IT', 'Marketing', 'Sales', 'IT', 'Finance', 'Marketing'], 
        'Age': [30, 40, 25, 35, 45, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'], 
        'Salary': [50000, 60000, 45000, 55000, 70000, 55000], 
        'Experience': [3, 7, 2, 5, 10, 4]}

employee_df = pd.DataFrame(data)

# Using iloc to select the first 3 rows
first_three_rows = employee_df.iloc[:3] 
print("First 3 rows:\n", first_three_rows)

# Using loc to select all rows where Department is "Marketing"
marketing_dept = employee_df.loc[employee_df['Department'] == 'Marketing']
print("\nMarketing Department:\n", marketing_dept)

# Using iloc to select Age and Gender columns for the first 4 rows
age_gender_first_four = employee_df.iloc[:4, [2, 3]] 
print("\nAge and Gender for first 4 rows:\n", age_gender_first_four)

# Using loc to select Salary and Experience columns for all rows where Gender is "Male"
salary_exp_male = employee_df.loc[employee_df['Gender'] == 'Male', ['Salary', 'Experience']]
print("\nSalary and Experience for Male employees:\n", salary_exp_male)
```
